## 1. Introducción a SQL Joins

### 1.1 ¿Qué es un JOIN?
En las bases de datos relacionales, los datos se distribuyen en múltiples tablas normalizadas para evitar redundancias y anomalías. Un `JOIN` (combinación) es una cláusula SQL que permite combinar columnas de una o más tablas basándose en una relación lógica entre ellas (generalmente una coincidencia de Clave Primaria y Clave Foránea).

### 1.2 El Modelo Relacional INFOTECH
Para estas consultas, utilizaremos la base de datos `INFOTECH`. Las relaciones clave son:
- `PERSONA(CI)` <-> `EMPLEADO(CI)` <-> `INGENIERO(CI)` / `ADMINISTRATIVO(CI)`
- `PERSONA(CI)` <-> `TELEFONO(CI)`
- `EMPLEADO(CI)` <-> `CLIENTE(CI_Representante)`
- `EMPLEADO(CI)` <-> `TRABAJO_EN(CI_Empleado)`
- `PROYECTO(Codigo)` <-> `TRABAJO_EN(Codigo_Proyecto)`
- `PROPIETARIO(ID_Propietario)` <-> `ES_PROPIETARIO(ID_Propietario)` / `EQUIPO_COMPUTACIONAL(ID_Propietario)`

## 2. INNER JOIN (Combinación Interna)

### 2.1 Concepto
El `INNER JOIN` devuelve únicamente las filas que tienen valores coincidentes en ambas tablas involucradas en la relación. Si una fila de la izquierda no tiene coincidencia en la de la derecha, esa fila no aparecerá en el resultado.

**Sintaxis general:**
```sql
SELECT tabla1.columna, tabla2.columna
FROM tabla1
INNER JOIN tabla2 ON tabla1.clave_comun = tabla2.clave_comun;
```

### 2.2 Ejemplos Demostrativos

#### Ejemplo 2.1: Consultar los nombres y apellidos de las personas que son empleados, junto con su salario base

````python
%%sql
SELECT p.Primer_Nombre, p.Primer_Apellido, e.Salario_Base
FROM PERSONA p
INNER JOIN EMPLEADO e ON p.CI = e.CI;
````

#### Ejemplo 2.2: Obtener las especialidades de los ingenieros, sus salarios y sus nombres completos

````python
%%sql
SELECT p.Primer_Nombre, p.Primer_Apellido, e.Salario_Base, i.Especialidad
FROM PERSONA p
INNER JOIN EMPLEADO e ON p.CI = e.CI
INNER JOIN INGENIERO i ON e.CI = i.CI;
````

## 3. LEFT JOIN (LEFT OUTER JOIN)

### 3.1 Concepto
El `LEFT JOIN` devuelve todas las filas de la tabla de la izquierda (especificada antes de la cláusula JOIN) y las filas coincidentes de la tabla de la derecha. Si no hay coincidencia, las columnas de la tabla de la derecha devolverán valores `NULL`.

### 3.2 Ejemplos Demostrativos

#### Ejemplo 3.1: Listar todas las personas de la base de datos y sus salarios bases si son empleados (si no lo son, se mostrará NULL)

````python
%%sql
SELECT p.CI, p.Primer_Nombre, p.Primer_Apellido, e.Salario_Base
FROM PERSONA p
LEFT JOIN EMPLEADO e ON p.CI = e.CI;
````

## 4. RIGHT JOIN (RIGHT OUTER JOIN)

### 4.1 Concepto
El `RIGHT JOIN` devuelve todas las filas de la tabla de la derecha (especificada después de la cláusula JOIN) y las filas coincidentes de la tabla de la izquierda. Es la inversa exacta de `LEFT JOIN`. Si no hay coincidencia en la izquierda, los resultados para esa tabla serán `NULL`.

### 4.2 Ejemplos Demostrativos

#### Ejemplo 4.1: Mostrar los salarios de los empleados relacionando la tabla PERSONA por la derecha

````python
%%sql
SELECT p.Primer_Nombre, p.Primer_Apellido, e.Salario_Base
FROM PERSONA p
RIGHT JOIN EMPLEADO e ON p.CI = e.CI;
````

## 5. FULL OUTER JOIN (Combinación Externa Completa) [INFORMATIVO - NO EVALUADO EN EXAMEN]

> [!NOTE]
> **Tema Informativo:** Este contenido sirve como referencia y desarrollo profesional. No se contempla en la evaluación práctica ni teórica del examen.

### 5.1 Concepto
El `FULL OUTER JOIN` devuelve todas las filas de la tabla izquierda y todas las de la derecha, coincidan o no. Si no hay coincidencia, las columnas de la tabla correspondiente tendrán valores `NULL`.

**Nota importante sobre MySQL:**
MySQL **no soporta** la cláusula `FULL OUTER JOIN` de manera nativa. Para emular un FULL JOIN en MySQL, se debe realizar un `LEFT JOIN` y un `RIGHT JOIN` combinando sus resultados mediante el operador `UNION` (el cual elimina los registros duplicados automáticamente).

### 5.2 Ejemplo Demostrativo
#### Ejemplo 5.1: Emular un FULL OUTER JOIN para obtener todos los empleados y todas las empresas de la base de datos

````python
%%sql
SELECT p.CI AS Persona_CI, p.Primer_Nombre, emp.Nombre AS Nombre_Empresa
FROM PERSONA p
LEFT JOIN ES_PROPIETARIO ep ON p.CI = ep.CI_Persona
LEFT JOIN EMPRESA emp ON ep.ID_Empresa = emp.ID_Empresa
UNION
SELECT p.CI AS Persona_CI, p.Primer_Nombre, emp.Nombre AS Nombre_Empresa
FROM PERSONA p
RIGHT JOIN ES_PROPIETARIO ep ON p.CI = ep.CI_Persona
RIGHT JOIN EMPRESA emp ON ep.ID_Empresa = emp.ID_Empresa;
````

## 6. SELF JOIN (Autocombinación) [INFORMATIVO - NO EVALUADO EN EXAMEN]

> [!NOTE]
> **Tema Informativo:** Este contenido sirve como referencia y desarrollo profesional. No se contempla en la evaluación práctica ni teórica del examen.

### 6.1 Concepto
Un `SELF JOIN` es una combinación normal, pero en la que una tabla se une consigo misma. Es muy útil para consultar datos jerárquicos o comparar filas dentro de la misma tabla. Para realizar un `SELF JOIN`, es obligatorio usar **alias diferentes** para referirse a la tabla en sus dos roles distintos.

### 6.2 Ejemplo Demostrativo
#### Ejemplo 6.1: Encontrar parejas de personas de la misma edad (mismo año de nacimiento) que tengan diferente cédula

````python
%%sql
SELECT p1.Primer_Nombre AS Persona_A, p2.Primer_Nombre AS Persona_B, YEAR(p1.Fecha_Nacimiento) AS Anio_Nacimiento
FROM PERSONA p1
INNER JOIN PERSONA p2 ON YEAR(p1.Fecha_Nacimiento) = YEAR(p2.Fecha_Nacimiento) AND p1.CI < p2.CI;
````

## 7. CROSS JOIN (Producto Cartesiano) [INFORMATIVO - NO EVALUADO EN EXAMEN]

> [!NOTE]
> **Tema Informativo:** Este contenido sirve como referencia y desarrollo profesional. No se contempla en la evaluación práctica ni teórica del examen.

### 7.1 Concepto
El `CROSS JOIN` devuelve el producto cartesiano de las filas de las tablas combinadas. Es decir, combina cada fila de la primera tabla con cada una de las filas de la segunda tabla. No requiere una condición lógica `ON`.

### 7.2 Ejemplo Demostrativo
#### Ejemplo 7.1: Combinar todas las marcas de equipos computacionales únicas con las especialidades de los ingenieros

````python
%%sql
SELECT DISTINCT eq.Marca, ing.Especialidad
FROM EQUIPO_COMPUTACIONAL eq
CROSS JOIN INGENIERO ing;
````

## 8. Combinaciones Múltiples (Unión de 3 o más tablas) [INFORMATIVO - NO EVALUADO EN EXAMEN]

> [!NOTE]
> **Tema Informativo:** Este contenido sirve como referencia y desarrollo profesional. No se contempla en la evaluación práctica ni teórica del examen.

### 8.1 Concepto
A menudo, la información que necesitamos está fragmentada en más de dos tablas. En SQL es común encadenar múltiples cláusulas `JOIN` secuenciales. Cada JOIN asocia la tabla resultante de los pasos anteriores con una nueva tabla.

### 8.2 Ejemplo Demostrativo
#### Ejemplo 8.1: Mostrar los nombres completos de las personas, su salario y el área del administrativo al que corresponden

````python
%%sql
SELECT p.Primer_Nombre, p.Primer_Apellido, e.Salario_Base, a.Area
FROM PERSONA p
INNER JOIN EMPLEADO e ON p.CI = e.CI
INNER JOIN ADMINISTRATIVO a ON e.CI = a.CI;
````
